**Problem Statement:** Analyze and visualize the Titanic dataset using Matplotlib and Seaborn. The dataset contains information about passengers, including their survival status, age, gender, and class.

*Tasks:*

Load the Titanic dataset into a Pandas DataFrame.

Visualize the distribution of passenger ages using a histogram.

Create a bar plot to compare survival rates by passenger class.

Use a heatmap to visualize the correlation between numerical features.

Create a pair plot to explore relationships between age, fare, and survival.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
#Load the Titanic dataset into a Pandas DataFrame.
data=sns.load_dataset('titanic')
df=pd.DataFrame(data)
df.drop_duplicates()
df.dropna()
df.head()
```

```python
#Visualize the distribution of passenger ages using a histogram.
plt.hist(df['age'], bins=20)
plt.xlabel('Pessanger')
plt.ylabel("Age")
plt.title('Pessanger Age Distribution')
plt.show()
```

```python
#Create a bar plot to compare survival rates by passenger class.
sns.barplot(x=df['class'], y=df['survived'], palette='Set1')
plt.xlabel('Class')
plt.ylabel('Survived')
plt.title('Survival rates by pessenger class')
plt.show()
```

```python
#Use a heatmap to visualize the correlation between numerical features.
data1=df[['survived','pclass','age','sibsp','parch','fare']]
corr=data1.corr()
sns.heatmap(corr, annot=True, cmap='coolwarm')
plt.show()
```

```python
#Create a pair plot to explore relationships between age, fare, and survival.
data2=df[['survived','age','fare']]
sns.pairplot(data=data2, hue='survived', palette='Set1')
plt.title("Pair Plot: Age, Fare, and Survival")
plt.show()
```
