**Matplotlib** is a powerful library for creating static, animated, and interactive visualizations in Python. It provides a MATLAB-like interface and is highly customizable.

*Key Features:*

Line Plots: For visualizing trends over time.

Bar Plots: For comparing categories.

Scatter Plots: For showing relationships between variables.

Histograms: For visualizing distributions.

```python
import matplotlib.pyplot as plt
x=[1,2,3,4,5]
y=[10,20,30,50,40]
#Line Plot
plt.plot(x,y, marker='o', linestyle='--', color='b', label='Line Plot')
plt.xlabel('X')
plt.ylabel('Y')
plt.title('Line Plot Example')
plt.show()
```

```python
#bar plot
z=['A', 'B', 'C', 'D', 'E']
plt.bar(z,y, color='g', label='Bar Plot')
plt.xlabel('X')
plt.ylabel('Y')
plt.title('Bar Plot Example')
plt.show()
```

```python
#Scatter plot
plt.scatter(x,y, color='r', marker='x', label='Scatter Plot')
plt.xlabel('X')
plt.ylabel('Y')
plt.title('Scatter Plot Example')
plt.show()
```

```python
#Histogram
# Data
data = [10, 15, 20, 25, 30, 35, 40, 45, 50]

# Create a histogram
plt.hist(data, bins=5, color='green', label='Histogram')
plt.xlabel("Values")
plt.ylabel("Frequency")
plt.title("Histogram Example")
plt.legend()
plt.show()
```

**Seaborn** is a high-level library built on top of Matplotlib. It provides beautiful default styles and simplifies the creation of complex visualizations.

*Key Features:*

Statistical Plots: Easily create plots like heatmaps, pair plots, and violin plots.

Styling: Beautiful default themes and color palettes.

Integration: Works seamlessly with Pandas DataFrames.

```python
import numpy as np
import seaborn as sns
data=np.random.rand(5,5)
sns.heatmap(data, annot=True, cmap='coolwarm')
plt.title("Heatmap example")
plt.show()
```

```python
#Pairplot
df=sns.load_dataset('iris')

sns.pairplot(df, hue='species')
plt.show()
```

```python
sns.violinplot(x='species', y='petal_length', data=df, palette='Set2')
plt.show()
```

```python
sns.boxplot(x='species', y='petal_width', data=df, palette='Set3')
plt.show()
```
