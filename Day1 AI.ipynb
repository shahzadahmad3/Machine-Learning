```python
# Base class (Animal)
class Animal:
    def __init__(self, name, age):
        self.name = name
        self.age = age
    
    def eat(self):
        print(f"{self.name} is eating...")

# Inheritance
# Derived class (Dog)
class Dog(Animal):
    def __init__(self, name, age):  # Define the __init__ method
        super().__init__(name, age)  # Call the parent constructor

    def eat(self):  # Polymorphism: Overriding the eat method
        print(f"{self.name} is munching on food.")

    def bark(self):
        print(f"{self.name} is barking.")

# Create an object of the Dog class
dog1 = Dog("Buddy", "12")

# Calling methods
dog1.eat()
dog1.bark()
```
