**Problem Statement:** Analyze a database of sales transactions. The database contains two tables:

*Customers:* customer_id, name, email

*Orders:* order_id, customer_id, product, amount

```python
import sqlite3

conn=sqlite3.connect('mydb.db')
cursor=conn.cursor()
```

```python
#Create the customers and orders tables.
cursor.execute("""CREATE TABLE IF NOT EXISTS customers(
               customer_id INTEGER PRIMARY KEY,
               name TEXT NOT NULL,
               email EMAIL
               )""")

cursor.execute("""
CREATE TABLE orders (
    order_id INTEGER PRIMARY KEY,
    customer_id INTEGER,
    product TEXT,
    amount REAL,
    FOREIGN KEY (customer_id) REFERENCES customers(customer_id)
)
""")
```

```python
#Insert sample data into both tables.
cursor.execute("""
INSERT INTO customers (name, email)
VALUES ('Alice', 'alice@example.com'),
       ('Bob', 'bob@example.com'),
       ('Charlie', 'charlie@example.com')
""")

cursor.execute("""INSERT INTO orders (customer_id, product, amount)
VALUES (1, 'Laptop', 1000),
       (1, 'Mouse', 20),
       (2, 'Keyboard', 50),
       (3, 'Monitor', 200),
       (3, 'Printer', 150)""")

# Commit changes
conn.commit()
```

```python
# Query 1: Retrieve all customers and their orders
cursor.execute("""
SELECT customers.name, orders.product, orders.amount
FROM customers
JOIN orders ON customers.customer_id = orders.customer_id
""")
print(cursor.fetchall())
```

```python
#Calculate the total revenue generated by each customer.
cursor.execute("""
SELECT customers.name, SUM(orders.amount) AS total_revenue
FROM customers
INNER JOIN orders ON customers.customer_id = orders.customer_id
               GROUP BY customers.name
""")
print(cursor.fetchall())
```

```python
#Find the top 3 customers by total spending
cursor.execute("""
    SELECT customers.name, SUM(orders.amount) AS total_revenue
    FROM customers
    INNER JOIN orders ON customers.customer_id = orders.customer_id
    GROUP BY customers.name
    ORDER BY total_revenue DESC
    LIMIT 3
""")
print(cursor.fetchall())
```

```python
# Close the connection
conn.close()
```
