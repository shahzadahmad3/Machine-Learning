**Probability** is the study of uncertainty. It quantifies how likely an event is to occur.

*Key Concepts:*

Experiment: A procedure with well-defined possible outcomes (e.g., rolling a die).
Sample Space (S): The set of all possible outcomes (e.g., S={1,2,3,4,5,6}).
Event (E): A subset of the sample space (e.g., rolling an even number).
Probability of an Event (P(E)):

P(E)= Total number of possible outcomes/Number of favorable outcomes

**Conditional Probability:** Probability of 𝐴 given 𝐵

P(A∣B)= P(A∩B)/P(B)
​


**Key Probability Rules**

*Addition Rule:*
P(A∪B)=P(A)+P(B)−P(A∩B)

*Multiplication Rule:*
P(A∩B)=P(A)P(B∣A) (Conditional Probability)

*Bayes’ theorem* describes the probability of an event based on prior knowledge of conditions related to the event.

P(A∣B)= P(B∣A).P(A)/P(B)

​

A **probability** distribution describes how probabilities are distributed over the values of a random variable.

*Key Distributions:*

Uniform Distribution: All outcomes are equally likely.

Normal Distribution (Gaussian): Symmetric distribution with a bell-shaped curve.

Binomial Distribution: Number of successes in a fixed number of trials.

```python
#Conditional Prabability
#Bionmial Probability
def binomial_probability(n, k ,p):
    from math import comb
    return comb(n,k)*(p**k)*(1-p)**(n-k)
def conditional_probabality(p_a_and_b, p_b):
    return p_a_and_b/p_b

n = 5  # Number of trials
k = 3  # Number of successes
p = 0.5  # Probability of success
print(f"Bionomial Probability of {n} success in {k} trial:", binomial_probability(n,k,p))
#Suppose you roll a fair six-sided die. What is the probability of rolling a 4 given that the roll is even?
p_a_and_b=1/6
p_b=3/6
print("Condition Probability:", conditional_probabality(p_a_and_b, p_b))
```
