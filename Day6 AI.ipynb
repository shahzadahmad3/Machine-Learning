```python
import sqlite3

# Connect to (or create) a database file
conn = sqlite3.connect('example.db')

# Create a cursor object
cursor = conn.cursor()
```

```python
# Create a table (if it doesn't exist)
cursor.execute("""
CREATE TABLE IF NOT EXISTS employees (
             id INTEGER PRIMARY KEY,  -- Fixed PRIMARY_KEY
             name TEXT NOT NULL,
             age INTEGER,
             department TEXT,
             salary REAL
             )
""")
```

```python
# Insert data
cursor.execute("INSERT INTO employees(name, age, department, salary) VALUES (?, ?, ?, ?)", 
               ("SHAHZAD", 40, "SED", 10000))  # Salary should be a number, not a string
cursor.execute("INSERT INTO employees(name, age, department, salary) VALUES (?, ?, ?, ?)", 
               ("FAROOQ", 45, "SED", 15000))

# Commit changes
conn.commit()
```

```python
#Retrieve specific columns
cursor.execute("SELECT age, department FROM employees")
print(cursor.fetchall())
#Filter data
cursor.execute("SELECT * FROM employees WHERE salary>10000")
print(cursor.fetchall())
#Sort data
cursor.execute("SELECT * FROM employees ORDER BY salary DESC")
print(cursor.fetchall())
```

```python
#Count the number of employees
cursor.execute("SELECT COUNT(*) FROM employees")
print(cursor.fetchall())
#Calculate average salary
cursor.execute("SELECT AVG(salary) FROM employees")
print(cursor.fetchall())
#Find the highest salary
cursor.execute("SELECT MAX(salary) FROM employees")
print(cursor.fetchall())
```

```python
#Deleting data
cursor.execute("DELETE FROM employees WHERE name='SHAHZAD'")

#Updating data
cursor.execute("UPDATE employees SET salary=20000 WHERE name='FAROOQ'")

# Fetch all records/Retrieve all data
cursor.execute("SELECT * FROM employees")
print(cursor.fetchall())
```

```python
#Grouping Data
#Average salary by department
cursor.execute("SELECT department, AVG(salary) FROM employees GROUP BY department")
print(cursor.fetchall())
```

```python
# Create the departments table (if it doesn't exist)
cursor.execute("""CREATE TABLE IF NOT EXISTS departments (
    id INTEGER PRIMARY KEY,
    name TEXT NOT NULL
)""")
```

```python
# Insert data into the 'departments' table
cursor.execute("INSERT INTO departments (name) VALUES (?)", ('SED',))
cursor.execute("INSERT INTO departments (name) VALUES (?)", ('HR',))
cursor.execute("INSERT INTO departments (name) VALUES (?)", ('Finance',))

# Commit changes
conn.commit()
cursor.execute("SELECT * FROM departments")
print(cursor.fetchall())
```

```python
#Join employees and departments
cursor.execute("SELECT employees.name, employees.salary, departments.name AS department FROM employees JOIN departments ON employees.department = departments.name")
print(cursor.fetchall())
```

```python
# Close the connection
conn.close()
```
