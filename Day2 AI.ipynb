**Lists**
A list is an ordered, mutable collection of items. Lists can store elements of different data types.

**Key Features:**
Ordered: Items have a defined order.

Mutable: Items can be added, removed, or modified.

Indexed: Items can be accessed using their index (starting from 0).

```python
list1=['Apple', 'Banana', 'Mango','Orange']
for i in list1:
    print(i)
```

```python
#Accessing Elements:
print(list1[1])
#Slicing:
print(list1[1:3])
# Appending(At the end):
list1.append('Strawberry')
# Removing:
list1.remove('Apple')
print(list1)
# List Comprehensions:
list2=[x**2 for x in range(5)]
print(list2)
```

**Tuples**
A tuple is an ordered, immutable collection of items. Once created, its elements cannot be changed.

**Key Features:**
Ordered: Items have a defined order.

Immutable: Items cannot be added, removed, or modified.

Indexed: Items can be accessed using their index.

```python
tuple1=(10,20,30, 40)
#Accessing Elements:
print(tuple1[1])
#Slicing:
print(tuple1[1:3])
print(tuple1[-1])
```

**Sets**
A set is an unordered collection of unique items. Sets are useful for operations like union, intersection, and difference.

**Key Features:**
Unordered: Items do not have a defined order.

Unique: No duplicate items allowed.

Mutable: Items can be added or removed.

```python
set1={1,2,3}
set2={4,5,6}
# Adding Items:
set2.add(7)
set2.remove(5)
print(set2)
print(set1.union(set2))
print(set1.intersection(set2))
print(set1.difference(set2))
```

**Dictionaries**
A dictionary is an unordered collection of key-value pairs. Keys must be unique and immutable (e.g., strings, numbers).

**Key Features:**

Unordered: Items do not have a defined order.

Key-Value Pairs: Each item is a pair of a key and its corresponding value.

Mutable: Items can be added, removed, or modified.

```python
dic1={'Name':'Shahzad', 'Age':'40', 'degree':'M.Phil'}
#Accessing Values:
print(dic1['Name'])
#Adding/Updating Items:
dic1['State']="Punjab"
print(dic1)
#Removing Items
dic1.pop('Age')
print(dic1)
#Dictionary Comprehensions
dic2={x:x**2 for x in range(5)}
print(dic2)
```
