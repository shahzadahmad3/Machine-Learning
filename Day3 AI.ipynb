**Generators** are a special type of function that allow you to iterate over a sequence of values without storing them all in memory at once. They use the yield keyword to produce values one at a time.

**Key Features:**

Memory-efficient: Generate values on-the-fly.

Lazy evaluation: Values are computed only when needed.

Can be iterated over using a for loop or next() function.

```python
def even_numbers():
    num=0
    while True:
        yield num
        num+=2
evens=even_numbers()
for _ in range(5):
    print(next(evens))
    print(next(evens))
```

```python
# Write a generator function to yield Fibonacci numbers up to a given limit.

# Use the next() function to manually fetch values from a generator

def fabonacci():
    a,b=0,1
    while True:
        yield a
        a,b=b, a+b # Update values
fib=fabonacci()
for _ in range(10):
    print(next(fib))
```

**Decorators** are functions that modify the behavior of other functions. They are often used for logging, timing, or adding functionality to existing functions.

**Key Features:**

Take a function as input and return a modified function.

Use the @decorator_name syntax to apply a decorator.

```python
import time

def time_decorator(func):
    def wrapper(*args, **kwargs):
        start_time=time.time()
        result=func(*args, **kwargs)
        end_time=time.time()
        print(f"The {func.__name__} takes time {end_time-start_time:.2f} seconds")
    return wrapper

@time_decorator
def slow_gen():
    time.sleep(2)
slow_gen()
```

**Multiprocessing** allows you to run multiple processes in parallel, taking advantage of multiple CPU cores. This is useful for CPU-bound tasks.

**Key Features:**

Uses the multiprocessing module.

Each process runs in its own memory space.

Processes can communicate using Queue or Pipe.

```python
from multiprocessing import Process

def square_number(n):
    """Function to print squares of numbers from 1 to 5."""
    for num in range(0, n):  # Loop through numbers 1 to 5
        print(f"Square of {num} is {num ** 2}")

# Creating two separate processes
p1 = Process(target=square_number, args=(5,)) # Pass arguments as a tuple (5,)
p2 = Process(target=square_number, args=(5,))

# Start the processes
p1.start()
p2.start()

# Wait for processes to complete
p1.join()
p2.join()
```

```python
with open('file.txt', 'w') as file:
    file.write("Hello World!")
    file.close()
```
